# Credit Default EDA — UCI Taiwan dataset

Exploratory analysis of the real cached dataset (30,000 accounts, 22.1% default rate).

**Run after `python scripts/download_data.py`.**

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().resolve() / 'src'))

import matplotlib.pyplot as plt
import pandas as pd

from credit_risk_modeling.config import DEFAULT_CONFIG
from credit_risk_modeling.data import load_raw, prepare
from credit_risk_modeling.features import build_features
```

```python
frame = prepare(load_raw(DEFAULT_CONFIG))
print(frame['default'].value_counts(normalize=True).round(4))
frame['limit_bal'].plot(kind='hist', bins=50, figsize=(8,3), title='Credit limit distribution')
plt.tight_layout(); plt.show()
```

## Default rate by repayment status (latest month)

The strongest single predictor: clients already 1+ month behind default far more often.

```python
rate = frame.groupby('pay_sep')['default'].agg(['mean', 'size'])
rate['mean'].plot(kind='bar', figsize=(8,3), color='tab:red', title='Default rate by pay_sep status')
plt.xlabel('repayment status (-2/-1/0 = current, 1..9 = months delay)'); plt.tight_layout(); plt.show()
print(rate.round(3))
```

## Takeaways

- Strong monotone relationship between payment delays and default - matches the permutation-importance results.
- 22% base rate: accuracy is a misleading headline metric; PR-AUC and cost-based thresholds are the right tools.
